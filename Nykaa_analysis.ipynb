```python
def iso_to_seconds(duration: str) -> int:
    m = re.fullmatch(
        r"P(?:(\d+)D)?(?:T(?:(\d+)H)?(?:(\d+)M)?(?:(\d+)S)?)?", duration
    )
    if not m:
        return 0
    d, h, mi, s = (int(x) if x else 0 for x in m.groups())
    return d * 86400 + h * 3600 + mi * 60 + s


def get_uploads_playlist_id(handle: str) -> str:
    resp = youtube.channels().list(part="contentDetails", forHandle=handle).execute()
    items = resp.get("items", [])
    if not items:
        raise ValueError(f"Channel not found for handle {handle}")
    return items[0]["contentDetails"]["relatedPlaylists"]["uploads"]


def get_video_ids(playlist_id: str, limit: int) -> list:
    ids, token = [], None
    while len(ids) < limit:
        resp = youtube.playlistItems().list(
            part="contentDetails", playlistId=playlist_id,
            maxResults=50, pageToken=token,
        ).execute()
        ids += [it["contentDetails"]["videoId"] for it in resp["items"]]
        token = resp.get("nextPageToken")
        if not token:
            break
    return ids[:limit]


def get_video_details(video_ids: list) -> pd.DataFrame:
    rows = []
    for i in range(0, len(video_ids), 50):
        chunk = video_ids[i:i + 50]
        resp = youtube.videos().list(
            part="snippet,contentDetails,statistics", id=",".join(chunk)
        ).execute()
        for v in resp["items"]:
            s, stats = v["snippet"], v.get("statistics", {})
            rows.append({
                "video_id": v["id"],
                "title": s["title"],
                "description": s.get("description", ""),
                "tags": ", ".join(s.get("tags", [])),
                "published": s["publishedAt"],
                "duration_s": iso_to_seconds(v["contentDetails"]["duration"]),
                "views": int(stats.get("viewCount", 0)),
                # likes/comments can be hidden by the channel -> stays empty (NaN)
                "likes": int(stats["likeCount"]) if "likeCount" in stats else None,
                "comments": int(stats["commentCount"]) if "commentCount" in stats else None,
            })
    return pd.DataFrame(rows)


def bucket_format(seconds: int) -> str:
    # The API doesn't flag Shorts. Shorts can be up to 3 minutes, so 61-180s is ambiguous.
    if seconds <= 60:
        return "Short (<=60s)"
    if seconds <= 180:
        return "Ambiguous (61-180s)"
    return "Long (>180s)"


def _has(text: str, kw: str) -> bool:
    return re.search(r"(?<!\w)" + re.escape(kw) + r"(?!\w)", text) is not None


def classify_method(title: str, description: str) -> str:
    text = f"{title} {description[:300]}".lower()
    for method, keywords in METHOD_RULES.items():
        if any(_has(text, kw) for kw in keywords):
            return method
    return "Other"


def add_features(df: pd.DataFrame) -> pd.DataFrame:
    df = df.copy()
    df["published"] = pd.to_datetime(df["published"], utc=True)
    for col in ("likes", "comments"):
        df[col] = pd.to_numeric(df[col], errors="coerce")

    df["format"] = df["duration_s"].apply(bucket_format)
    df["method"] = [classify_method(t, d) for t, d in zip(df["title"], df["description"])]

    # Engagement rate (%) = (likes + comments) / views
    df["engagement_rate"] = ((df["likes"] + df["comments"].fillna(0)) / df["views"] * 100)
    df.loc[df["views"] == 0, "engagement_rate"] = float("nan")

    # Views per day, so older videos don't win just by being older
    now = pd.Timestamp.now(tz="UTC")
    days = ((now - df["published"]).dt.total_seconds() / 86400).clip(lower=1)
    df["views_per_day"] = df["views"] / days

    # Description features
    df["desc_length"] = df["description"].str.len()
    df["hashtag_count"] = df["description"].apply(lambda d: len(re.findall(r"#\w+", d)))
    df["has_link"] = df["description"].str.contains("http", case=False)
    df["has_cta"] = df["description"].str.lower().apply(
        lambda d: any(_has(d, w) for w in CTA_WORDS)
    )
    return df


def summarize(df: pd.DataFrame, by) -> pd.DataFrame:
    out = df.groupby(by).agg(
        videos=("video_id", "count"),
        median_views=("views", "median"),
        median_views_per_day=("views_per_day", "median"),
        median_engagement_pct=("engagement_rate", "median"),
    ).round(2)
    out["small_sample"] = out["videos"] < MIN_GROUP
    return out.sort_values("median_views_per_day", ascending=False)


def top_hashtags(df: pd.DataFrame, n: int = 15):
    top = df[df["views"] >= df["views"].quantile(0.75)]
    counts = Counter(h.lower() for d in top["description"] for h in re.findall(r"#\w+", d))
    return counts.most_common(n)


def main():
    global youtube
    youtube = build("youtube", "v3", developerKey=API_KEY)

    playlist_id = get_uploads_playlist_id(CHANNEL_HANDLE)
    df = get_video_details(get_video_ids(playlist_id, MAX_VIDEOS))
    df = add_features(df)

    if RECENT_DAYS:
        cutoff = pd.Timestamp.now(tz="UTC") - pd.Timedelta(days=RECENT_DAYS)
        df = df[df["published"] >= cutoff]

    print(f"\nVideos analysed: {len(df)}\n")
    print("=== By METHOD (sorted by median views per day) ===")
    print(summarize(df, "method"))
    print("\n=== By FORMAT ===")
    print(summarize(df, "format"))
    print("\n=== Median views: METHOD x FORMAT ===")
    print(df.pivot_table(index="method", columns="format", values="views", aggfunc="median"))
    print("\n=== Description features (median views/day) ===")
    print(df.groupby("has_cta")["views_per_day"].median().round(2))
    print(df.groupby("has_link")["views_per_day"].median().round(2))
    print("\n=== Most common hashtags in the top-25% videos ===")
    for tag, c in top_hashtags(df):
        print(f"{tag}: {c}")

    # df.sort_values("views", ascending=False).to_csv("channel_videos_full.csv", index=False)
    # summarize(df, "method").to_csv("method_summary.csv")
    # print("\nSaved channel_videos_full.csv and method_summary.csv")
    # print("TIP: open channel_videos_full.csv and manually check ~30 'method' labels.")


if __name__ == "__main__":
    main()
```
